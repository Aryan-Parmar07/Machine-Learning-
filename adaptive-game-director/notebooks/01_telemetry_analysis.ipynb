# 01. Gameplay Telemetry Analysis
**Project:** Adaptive Game Director (RL-Based DDA for ViZDoom)

This notebook analyzes the raw and derived telemetry collected during human gameplay sessions (`data/telemetry/human_session.csv`).
Key objectives:
1. Verify telemetry integrity and continuity across game tics.
2. Inspect temporal dynamics of player health, ammo, and kills.
3. Analyze correlation between combat intensity, enemy density, and player stress.

```python
import os
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Set project root
PROJECT_ROOT = Path("..").resolve()
csv_path = PROJECT_ROOT / "data" / "telemetry" / "human_session.csv"

print(f"Telemetry file path: {csv_path}")
print(f"File exists: {csv_path.exists()}")
```

```python
# Load session data
if csv_path.exists():
    df = pd.read_csv(csv_path)
    print(f"Loaded {len(df)} telemetry frames.")
    display(df.head())
else:
    print("No telemetry file found yet. Run `python scripts/play_human.py` first.")
```

```python
# Summary statistics
if csv_path.exists():
    summary_cols = ['health', 'armor', 'ammo', 'killcount', 'deathcount', 'player_speed', 'accuracy_estimate', 'player_skill_estimate']
    display(df[summary_cols].describe())
```

```python
# Plot Player Health & Ammo over time
if csv_path.exists():
    plt.figure(figsize=(12, 5))
    plt.plot(df['elapsed_time_sec'], df['health'], label='Health', color='red', lw=2)
    plt.plot(df['elapsed_time_sec'], df['ammo'], label='Ammo', color='blue', lw=1.5, alpha=0.8)
    plt.axhline(35, color='orange', linestyle='--', alpha=0.7, label='Target Health Lower Bound')
    plt.axhline(75, color='green', linestyle='--', alpha=0.7, label='Target Health Upper Bound')
    plt.title('Player Vitals over Time (Flow Channel Target: 35-75 HP)')
    plt.xlabel('Elapsed Time (s)')
    plt.ylabel('Value')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```

```python
# Plot Combat Dynamics (Damage Dealt vs Damage Taken)
if csv_path.exists():
    fig, ax1 = plt.subplots(figsize=(12, 5))
    ax1.plot(df['elapsed_time_sec'], df['damagecount'], label='Damage Dealt', color='crimson')
    ax1.plot(df['elapsed_time_sec'], df['damage_taken'], label='Damage Taken', color='darkorange')
    ax1.set_xlabel('Elapsed Time (s)')
    ax1.set_ylabel('Cumulative Damage')
    ax1.legend(loc='upper left')
    
    ax2 = ax1.twinx()
    ax2.plot(df['elapsed_time_sec'], df['enemy_count'], label='Enemy Count', color='purple', linestyle=':')
    ax2.set_ylabel('Active Enemies')
    ax2.legend(loc='upper right')
    
    plt.title('Combat Intensity vs Enemy Density')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```
