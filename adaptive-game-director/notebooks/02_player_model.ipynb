This notebook evaluates:

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

PROJECT_ROOT = Path("..").resolve()
sys.path.insert(0, str(PROJECT_ROOT))

from src.player_model.features import FeatureExtractor, RawTelemetryFrame
from src.player_model.skill_estimator import SkillEstimator

csv_path = PROJECT_ROOT / "data" / "telemetry" / "human_session.csv"
print(f"Telemetry data ready: {csv_path.exists()}")
```

```python
# Inspect Skill Estimate Trajectory over Time
if csv_path.exists():
    df = pd.read_csv(csv_path)
    plt.figure(figsize=(10, 4))
    plt.plot(df['elapsed_time_sec'], df['player_skill_estimate'], label='Skill Estimate', color='teal', lw=2)
    plt.axhline(0.35, color='orange', linestyle='--', label='Novice Threshold (0.35)')
    plt.axhline(0.75, color='darkgreen', linestyle='--', label='Expert Threshold (0.75)')
    plt.ylim(0.0, 1.0)
    plt.xlabel('Time (s)')
    plt.ylabel('Estimated Skill Score [0, 1]')
    plt.title('Player Skill Adaptation Curve')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()
```

```python
# Aggression Score vs Survival Dynamics
if csv_path.exists():
    plt.figure(figsize=(10, 4))
    plt.scatter(df['aggression_score'], df['kill_rate'], c=df['health'], cmap='viridis', alpha=0.7)
    cbar = plt.colorbar()
    cbar.set_label('Health')
    plt.xlabel('Aggression Score')
    plt.ylabel('Kill Rate (kills/sec)')
    plt.title('Player Aggression vs Kill Efficiency')
    plt.grid(True, alpha=0.3)
    plt.show()
```
